# Data Science starter

A small analysis with **pandas** and **matplotlib**:

1. load a table (the sample `sales.csv` next to this notebook, or built in code)
2. explore it (`head`, `describe`, group by)
3. plot it inline

Run a cell with **Shift+Enter**, everything with **Run All**.

```python
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_rows", 20)
print("pandas", pd.__version__)
```

```python
from pathlib import Path

if Path("sales.csv").exists():
    df = pd.read_csv("sales.csv")
else:
    df = pd.DataFrame({
        "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
        "region": ["North", "South", "North", "South", "North", "South"],
        "sales": [120, 98, 143, 110, 160, 131],
        "cost": [80, 70, 90, 75, 101, 88],
    })
df["profit"] = df["sales"] - df["cost"]
df.head()
```

```python
df.describe()
```

```python
df.groupby("region")[["sales", "profit"]].sum()
```

## Plot

Figures are shown under the cell (matplotlib *inline*).

```python
fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(df["month"], df["sales"], marker="o", label="sales")
ax.bar(df["month"], df["profit"], alpha=0.4, label="profit")
ax.set_title("Sales and profit")
ax.legend()
plt.show()
```
